```python
import os
import numpy as np
import pandas as pd
from PIL import Image
from tqdm import tqdm
import matplotlib.pyplot as plt

import torch
import torch.nn as nn
import torchvision.transforms as transforms
from torchvision.datasets import ImageFolder
from torch.utils.data import Dataset, DataLoader
from torchvision.utils import make_grid

# --------------------
# 1. Config
# --------------------
IMG_SIZE = 64
BATCH_SIZE = 64
Z_DIM = 100
EPOCHS = 50
DEVICE = torch.device("cuda" if torch.cuda.is_available() else "cpu")
DATA_DIR = "./dataset/celeba"  # path to CelebA images
ATTR_PATH = "./dataset/list_attr_celeba.txt"  # path to attributes

# --------------------
# 2. Dataset & Dataloader
# --------------------
class CelebADataset(Dataset):
    def __init__(self, img_dir, attr_path, transform=None):
        self.img_dir = img_dir
        self.transform = transform

        # Read attribute file
        df = pd.read_csv(attr_path, skiprows=1, delim_whitespace=True)
        df = df.reset_index().rename(columns={'index': 'filename'})
        df['LongHair'] = df['Male'].apply(lambda x: 0 if x == 1 else 1)  # Treat Female as Long Hair

        self.img_names = df['filename'].values
        self.labels = df['LongHair'].values

    def __len__(self):
        return len(self.img_names)

    def __getitem__(self, index):
        img_path = os.path.join(self.img_dir, self.img_names[index])
        image = Image.open(img_path).convert("RGB")

        label = self.labels[index]
        if self.transform:
            image = self.transform(image)

        return image, label


transform = transforms.Compose([
    transforms.Resize(IMG_SIZE),
    transforms.CenterCrop(IMG_SIZE),
    transforms.ToTensor(),
    transforms.Normalize([0.5]*3, [0.5]*3)
])

dataset = CelebADataset(DATA_DIR, ATTR_PATH, transform=transform)
dataloader = DataLoader(dataset, batch_size=BATCH_SIZE, shuffle=True)

# --------------------
# 3. Conditional GAN
# --------------------
class Generator(nn.Module):
    def __init__(self, z_dim, label_dim, img_channels=3, feature_g=64):
        super().__init__()
        self.label_emb = nn.Embedding(2, z_dim)
        self.model = nn.Sequential(
            nn.ConvTranspose2d(z_dim, feature_g * 8, 4, 1, 0, bias=False),
            nn.BatchNorm2d(feature_g * 8),
            nn.ReLU(True),

            nn.ConvTranspose2d(feature_g * 8, feature_g * 4, 4, 2, 1, bias=False),
            nn.BatchNorm2d(feature_g * 4),
            nn.ReLU(True),

            nn.ConvTranspose2d(feature_g * 4, feature_g * 2, 4, 2, 1, bias=False),
            nn.BatchNorm2d(feature_g * 2),
            nn.ReLU(True),

            nn.ConvTranspose2d(feature_g * 2, feature_g, 4, 2, 1, bias=False),
            nn.BatchNorm2d(feature_g),
            nn.ReLU(True),

            nn.ConvTranspose2d(feature_g, img_channels, 4, 2, 1, bias=False),
            nn.Tanh()
        )

    def forward(self, z, labels):
        label_embedding = self.label_emb(labels)
        x = z * label_embedding
        x = x.view(x.size(0), x.size(1), 1, 1)
        return self.model(x)


class Discriminator(nn.Module):
    def __init__(self, img_channels=3, label_dim=1, feature_d=64):
        super().__init__()
        self.label_emb = nn.Embedding(2, IMG_SIZE * IMG_SIZE)
        self.model = nn.Sequential(
            nn.Conv2d(img_channels + 1, feature_d, 4, 2, 1),
            nn.LeakyReLU(0.2, inplace=True),

            nn.Conv2d(feature_d, feature_d * 2, 4, 2, 1),
            nn.BatchNorm2d(feature_d * 2),
            nn.LeakyReLU(0.2, inplace=True),

            nn.Conv2d(feature_d * 2, feature_d * 4, 4, 2, 1),
            nn.BatchNorm2d(feature_d * 4),
            nn.LeakyReLU(0.2, inplace=True),

            nn.Conv2d(feature_d * 4, 1, 4, 1, 0),
            nn.Sigmoid()
        )

    def forward(self, x, labels):
        batch_size = x.size(0)
        label_embedding = self.label_emb(labels).view(batch_size, 1, IMG_SIZE, IMG_SIZE)
        x = torch.cat([x, label_embedding], dim=1)
        return self.model(x)

# --------------------
# 4. Initialize
# --------------------
G = Generator(Z_DIM, label_dim=1).to(DEVICE)
D = Discriminator().to(DEVICE)
criterion = nn.BCELoss()
opt_G = torch.optim.Adam(G.parameters(), lr=2e-4, betas=(0.5, 0.999))
opt_D = torch.optim.Adam(D.parameters(), lr=2e-4, betas=(0.5, 0.999))

# --------------------
# 5. Training Loop
# --------------------
print("Start Training...")
for epoch in range(EPOCHS):
    for real_imgs, labels in tqdm(dataloader):
        real_imgs, labels = real_imgs.to(DEVICE), labels.to(DEVICE)

        # Train Discriminator
        noise = torch.randn(real_imgs.size(0), Z_DIM).to(DEVICE)
        fake_imgs = G(noise, labels)

        real_validity = D(real_imgs, labels)
        fake_validity = D(fake_imgs.detach(), labels)

        real_labels = torch.ones_like(real_validity)
        fake_labels = torch.zeros_like(fake_validity)

        loss_D = (criterion(real_validity, real_labels) + criterion(fake_validity, fake_labels)) / 2

        opt_D.zero_grad()
        loss_D.backward()
        opt_D.step()

        # Train Generator
        gen_validity = D(fake_imgs, labels)
        loss_G = criterion(gen_validity, real_labels)

        opt_G.zero_grad()
        loss_G.backward()
        opt_G.step()

    print(f"Epoch [{epoch+1}/{EPOCHS}] | Loss_D: {loss_D.item():.4f} | Loss_G: {loss_G.item():.4f}")

    # Show sample
    if (epoch + 1) % 10 == 0:
        with torch.no_grad():
            test_z = torch.randn(16, Z_DIM).to(DEVICE)
            test_labels = torch.ones(16).long().to(DEVICE)  # Long hair
            test_images = G(test_z, test_labels)
            test_images = (test_images + 1) / 2
            grid = make_grid(test_images.cpu(), nrow=4)
            plt.figure(figsize=(8, 8))
            plt.imshow(grid.permute(1, 2, 0))
            plt.title(f"Generated Long Hair Faces (Epoch {epoch+1})")
            plt.axis("off")
            plt.show()
```
